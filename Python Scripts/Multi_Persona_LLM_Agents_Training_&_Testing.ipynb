Training Medical Model

```python
import torch
from datasets import load_dataset
from transformers import (
    AutoModelForCausalLM,
    AutoTokenizer,
    BitsAndBytesConfig,
    TrainingArguments,
    LlamaTokenizer, # Use LlamaTokenizer for OpenLLaMA
)
from peft import LoraConfig, get_peft_model, prepare_model_for_kbit_training
from trl import SFTTrainer

# --- 1. Define Model and Dataset ---
model_id = "openlm-research/open_llama_7b_v2"
dataset_id = "keivalya/MedQuad-MedicalQnADataset"
new_model_name = "openllama-7b-medquad-qlora" # Name for your fine-tuned model

# --- 2. QLoRA and Quantization Configuration ---
# 4-bit quantization configuration
bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_use_double_quant=True,
    bnb_4bit_quant_type="nf4",
    bnb_4bit_compute_dtype=torch.bfloat16, # Use bfloat16 for computation on supported GPUs
)

# LoRA configuration
lora_config = LoraConfig(
    r=64, # LoRA attention dimension
    lora_alpha=16, # Alpha parameter for LoRA scaling
    lora_dropout=0.1, # Dropout probability for LoRA layers
    bias="none",
    task_type="CAUSAL_LM",
)

# --- 3. Load Model and Tokenizer ---
print(f"Loading model: {model_id}...")
model = AutoModelForCausalLM.from_pretrained(
    model_id,
    quantization_config=bnb_config,
    device_map="auto",
    # ADD THIS LINE:
    use_safetensors=True
)

# OpenLLaMA V2 uses LlamaTokenizer (and requires use_fast=False sometimes)
# The LlamaTokenizer is part of the transformers library
tokenizer = LlamaTokenizer.from_pretrained(model_id)
tokenizer.pad_token = tokenizer.eos_token
tokenizer.padding_side = "right" # Fixes overflow issue in FP16

# Prepare the model for k-bit training
model = prepare_model_for_kbit_training(model)

# Apply LoRA config
model = get_peft_model(model, lora_config)

# --- 4. Load and Preprocess Dataset ---
print(f"Loading dataset: {dataset_id}...")
dataset = load_dataset(dataset_id, split="train")

# The MedQuad-MedicalQnADataset has 'Question' and 'Answer' columns.
# We need to format these into an instruction-following prompt template.
def formatting_function(examples):
    text = []
    for question, answer in zip(examples["Question"], examples["Answer"]):
        # Define the instruction prompt template
        prompt = f"### Question:\n{question}\n\n### Answer:\n{answer}{tokenizer.eos_token}"
        text.append(prompt)
    return {"text": text}

# Apply the formatting to the dataset
dataset = dataset.map(formatting_function, batched=True)

# --- 5. Training Arguments ---
training_arguments = TrainingArguments(
    output_dir=f"./results/{new_model_name}",
    num_train_epochs=3,                     # Number of training epochs
    per_device_train_batch_size=4,          # Batch size per GPU
    gradient_accumulation_steps=4,          # Increase effective batch size to 16 (4*4)
    optim="paged_adamw_8bit",               # Use 8-bit Adam optimizer for memory efficiency
    save_steps=500,                         # Save checkpoint every 500 steps
    logging_steps=50,                       # Log loss every 50 steps
    learning_rate=2e-4,                     # Learning rate
    weight_decay=0.001,
    fp16=False,                             # Set to True if your GPU supports it, but bfloat16 is preferred
    bf16=True,                              # Use bfloat16 precision
    max_grad_norm=0.3,
    max_steps=-1,                           # -1 means run for num_train_epochs
    warmup_ratio=0.03,
    group_by_length=True,
    lr_scheduler_type="constant",
)

# --- 6. Initialize and Train SFTTrainer ---
print("Starting SFTTrainer...")
max_seq_length = 1024 # Adjust based on GPU memory

trainer = SFTTrainer(
    model=model,
    train_dataset=dataset,
    peft_config=lora_config,
    dataset_text_field="text", # The column name containing the formatted text
    max_seq_length=max_seq_length,
    tokenizer=tokenizer,
    args=training_arguments,
    packing=False, # Set to True to pack short examples for efficiency
)

trainer.train()

# --- 7. Save the Fine-tuned Model ---
trainer.model.save_pretrained(new_model_name)
tokenizer.save_pretrained(new_model_name)
print(f"Fine-tuned model saved to: {new_model_name}")
```

```python
import torch
import os
from transformers import (
    AutoModelForCausalLM,
    AutoTokenizer,
    BitsAndBytesConfig
)
from peft import PeftModel

# --- 1. Configuration (MUST MATCH TRAINING) ---
# Check your environment for bfloat16 compatibility (e.g., NVIDIA A100, H100, RTX 30/40 series).
# Use torch.float16 (fp16) if bfloat16 (bf16) is not supported.

# The original base model ID
model_id = "openlm-research/open_llama_7b_v2"

# Path to your fine-tuned adapter weights
new_model_name = "openllama-7b-medquad-qlora" # Change this if your output name was different
adapter_path = f"./results/{new_model_name}/checkpoint-3078"

# Quantization configuration (must match the config used during training)
bnb_config = BitsAndBytesConfig(
    load_in_4bit=True,
    bnb_4bit_quant_type="nf4",  # Normal Float 4
    bnb_4bit_compute_dtype=torch.bfloat16, # Or torch.float16 if bf16 is unavailable
    bnb_4bit_use_double_quant=True,
)

# Set device to GPU 0
device_map = {"": 0}
torch_dtype_to_use = torch.bfloat16 # Match bnb_4bit_compute_dtype
# ---------------------------------------------

# --- 2. Load Base Model, Tokenizer, and Adapters ---

# Load Base Model (in 4-bit)
print("Loading base model with 4-bit quantization...")
try:
    base_model = AutoModelForCausalLM.from_pretrained(
        model_id,
        quantization_config=bnb_config,
        device_map=device_map,
        torch_dtype=torch_dtype_to_use,
    )
except Exception as e:
    print(f"Error loading base model: {e}")
    print("Ensure you have PyTorch, bitsandbytes, and transformers installed correctly.")
    exit()

# Load Tokenizer (using use_fast=False as required for LLaMA-based models sometimes)
print("Loading tokenizer...")
tokenizer = AutoTokenizer.from_pretrained(model_id, use_fast=False)
# Set pad token to EOS token to ensure generation stops correctly
tokenizer.pad_token = tokenizer.eos_token

# Attach LoRA Adapters
print(f"Loading and merging adapter weights from: {adapter_path}...")
try:
    # This loads the adapter weights on top of the 4-bit base model
    model = PeftModel.from_pretrained(base_model, adapter_path)
    # Set model to evaluation mode
    model.eval()
    print("Model and adapters loaded successfully.")
except Exception as e:
    print(f"Error loading PEFT model: {e}")
    print(f"Ensure the adapter path '{adapter_path}' exists and contains 'adapter_model.bin' and 'adapter_config.json'.")
    exit()

# --- 3. Inference Function with ATTENTION MASK FIX ---

def generate_response(prompt: str, max_new_tokens: int = 256):
    """
    Generates a response from the fine-tuned model, including the attention_mask.
    """

    # 1. Format the prompt (MUST match the instruction format used in training!)
    # Example format for instruction tuning:
    formatted_prompt = f"### Instruction:\n{prompt}\n\n### Response:"

    # 2. Tokenize the input (gets input_ids AND attention_mask)
    inputs = tokenizer(formatted_prompt, return_tensors="pt")

    # 3. Move BOTH tensors to the correct device
    input_ids = inputs["input_ids"].to(device_map[""])
    attention_mask = inputs["attention_mask"].to(device_map[""])

    # 4. Generate the response
    with torch.no_grad():
        outputs = model.generate(
            input_ids=input_ids,
            attention_mask=attention_mask,  # <--- FIX: Passing the attention mask
            max_new_tokens=max_new_tokens,
            do_sample=True,
            temperature=0.7,
            top_p=0.9,
            eos_token_id=tokenizer.eos_token_id,
        )

    # 5. Decode the output
    response_text = tokenizer.decode(outputs[0], skip_special_tokens=True)

    # 6. Clean up the output (extract only the model's response part)
    if "### Response:" in response_text:
        return response_text.split("### Response:")[1].strip()

    return response_text.strip() # Fallback

# --- 4. Test the Model ---

# Test Prompt 1: General Knowledge (Should be good due to base model)
general_prompt = "What is the capital of France?"

# Test Prompt 2: Domain-Specific (Should show fine-tuning effect)
# Replace this with a prompt related to your fine-tuning data (e.g., medical query for MedQuAD)
domain_prompt = "What are the common side effects of the drug Metformin?"

print("\n" + "=" * 80)
print(f"TESTING FINE-TUNED MODEL: {new_model_name}")
print("=" * 80)

# Test 1
print(f"\nPROMPT (General):\n{general_prompt}")
response_1 = generate_response(general_prompt, max_new_tokens=50)
print("-" * 80)
print(f"GENERATED RESPONSE:\n{response_1}")
print("-" * 80)

# Test 2
print(f"\nPROMPT (Domain-Specific):\n{domain_prompt}")
response_2 = generate_response(domain_prompt)
print("-" * 80)
print(f"GENERATED RESPONSE:\n{response_2}")
print("=" * 80)
```
